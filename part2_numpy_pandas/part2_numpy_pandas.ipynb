```python
us_views = views[:40949]
gb_views = views[40949:79865]
in_views = views[79865:]

print("US:", us_views.shape, us_views.max(), us_views.mean())
print("GB:", gb_views.shape, gb_views.max(), gb_views.mean())
print("IN:", in_views.shape, in_views.max(), in_views.mean())
```

```python
total_reactions = likes + dislikes
like_ratio = np.divide(likes, total_reactions, out=np.zeros(len(likes)), where=total_reactions != 0)

print(like_ratio[:5])
print("Min:", like_ratio.min(), "Max:", like_ratio.max())
```

```python
comment_rate = np.divide(comments, views, out=np.zeros(len(comments)), where=views != 0)

print(comment_rate[:5])
print("Min:", comment_rate.min(), "Max:", comment_rate.max())
```

```python
views_norm = (views - views.min()) / (views.max() - views.min())

print(views_norm[:35])
print("Min:", views_norm.min(), "Max:", views_norm.max())
```

```python
likes_norm = (likes - likes.min()) / (likes.max() - likes.min())

print("Min:", likes_norm.min(), "Max:", likes_norm.max())
print(likes_norm[:10])
```

```python
comments_norm = (comments - comments.min()) / (comments.max() - comments.min())

print("Min:", comments_norm.min(), "Max:", comments_norm.max())
print(comments_norm[:10])
```

## Min-max normalization
Views, likes and comments are on very different scales (views in millions, likes in thousands, comments in hundreds). If added directly, views would dominate the score. So each column was min-max normalized to a 0-1 scale using (value - min) / (max - min), which makes them comparable before combining them into the engagement score.

```python
engagement = 0.2 * views_norm + 0.3 * likes_norm + 0.5 * comments_norm

print(engagement[:5])
print("Min:", engagement.min(), "Max:", engagement.max())
```

```python
print(views[0], likes[0], comments[0])
print(views_norm[0], likes_norm[0], comments_norm[0])
print(engagement[0])
```

```python
median_score = np.median(engagement)
mean_score = engagement.mean()

print("Median:", median_score)
print("Mean:", mean_score)
```

## Mean vs median of engagement score
Median = 0.00145 and mean = 0.00821. The mean is much higher than the median, so the engagement score is right-skewed: a few very engaging videos pull the average up, while most videos have low scores. The median describes a typical video better here.



```python
cutoff = np.percentile(engagement, 90)
top10_mask = engagement >= cutoff

print("Cutoff:", cutoff)
print("Top 10% count:", top10_mask.sum())
```

## Top 10% by engagement
A boolean mask (engagement >= 90th percentile) flags the top 10% of records. Cutoff = 0.0182, and 11,722 of 117,217 records fall in the top 10%. A percentile is used instead of the mean because the scores are right-skewed.

```python
log_views = np.log10(views + 1)

print(log_views[:5])
print("Min:", log_views.min(), "Max:", log_views.max())
```

## Log-scaled views
Raw views are very skewed (a few videos have hundreds of millions, most have far fewer), so a plot of raw views is squeezed to one side. log10(views + 1) compresses the range, and the +1 avoids log(0).



```python
all_df["engagement"] = engagement
all_df["log_views"] = log_views
print(all_df.shape)
```

```python
all_df["trending_date"] = pd.to_datetime(all_df["trending_date"], format="%y.%d.%m")

counts = all_df.groupby(["trending_date", "country"]).size().unstack(fill_value=0)
print(counts.head())
print(counts.columns.tolist())

arr = counts.to_numpy()
print("Shape (days x countries):", arr.shape)
```

```python
flat = arr.flatten()
print("Flattened:", flat.shape)

reshaped = flat.reshape(205, 3)
print("Back to:", reshaped.shape)
print("Same as original?", (reshaped == arr).all())
```

```python
day_totals = arr.sum(axis=1, keepdims=True)
share = arr / day_totals

print("day_totals shape:", day_totals.shape)
print("share shape:", share.shape)
print(share[:5].round(3))
```

```python
country_avg = arr.mean(axis=0)
relative = arr / country_avg

print("country_avg:", country_avg.round(1))
print("country_avg shape:", country_avg.shape)
print(relative[:5].round(2))
```

## Reshape and broadcasting
Built a days x countries array of daily trending counts (205 x 3). flatten() made it 1D and reshape() restored it. Broadcasting divided each row by its day total (country share, about 0.33 each) and each column by its country average (relative value, 1.0 = average day). No loops were used.
