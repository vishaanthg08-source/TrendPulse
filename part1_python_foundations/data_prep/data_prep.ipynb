```python
!pip install -q kagglehub
import kagglehub, os, json
import pandas as pd

src = kagglehub.dataset_download("datasnaek/youtube-new")
print(os.listdir(src))
```

```python
os.makedirs("data", exist_ok=True)
for c in ["US", "GB", "IN"]:
    df = pd.read_csv(f"{src}/{c}videos.csv", encoding="utf-8", encoding_errors="replace")
    print(c, "original shape:", df.shape)
    df = df.drop(columns=["description", "tags", "thumbnail_link"], errors="ignore")
    df.to_csv(f"data/{c}videos.csv", index=False)
    print(c, "trimmed shape:", df.shape)
    print(c, "size MB:", round(os.path.getsize(f"data/{c}videos.csv") / 1e6, 2))
```

```python
with open(f"{src}/US_category_id.json") as f:
    j = json.load(f)

cats = pd.DataFrame(
    [{"category_id": int(i["id"]), "category_name": i["snippet"]["title"]} for i in j["items"]]
)
cats.to_csv("data/categories.csv", index=False)
print(cats.shape)
print(cats.head())
```
